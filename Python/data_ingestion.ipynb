```python
import os
import pandas as pd
from sqlalchemy import create_engine

# 1. Config
DB_USER = "root"
DB_PASSWORD = "password"
DB_HOST = "localhost"
DB_PORT = "3306"
DB_NAME = "UberData"
TABLE_NAME = "bookings"
CSV_PATH = r"C:\Users\HP-PC\Desktop\Project_X\Data\cleaned_booking.csv"

# 2. Read CSV
print(f"Reading: {CSV_PATH}")
df = pd.read_csv(CSV_PATH)
print(f"Loaded {len(df):,} rows from CSV.")

# 3. Connect & Append into existing SQL table
engine = create_engine(
    f"mysql+pymysql://{DB_USER}:{DB_PASSWORD}@{DB_HOST}:{DB_PORT}/{DB_NAME}"
)

print(f"Appending data into `{DB_NAME}`.`{TABLE_NAME}`...")
df.to_sql(
    name=TABLE_NAME,
    con=engine,
    if_exists="append",  # Inserts into the MySQL table you created
    index=False,
    chunksize=10000,
)

print(
    f"Successfully inserted {len(df):,} records into `{DB_NAME}`.`{TABLE_NAME}`!"
)
```
